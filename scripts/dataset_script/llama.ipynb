```python
client = Groq()
```

```python
def generate_content(topic):  
    response = client.chat.completions.create(
      messages = [
          {
              "role": "system",
              "content": f'''what do u think about {topic}, write 250 words about it, should have single paragraph'''
              },
          # {
          #     "role": "user",
          #     "content" : f'topic: {tuple(map(tuple, topic))}'
          #       }
                ],
      model = "llama-3.1-8b-instant",
      temperature=1,
      max_tokens=1024,
      top_p=1
    )
    return response
```

```python
def find_second_maximum(numbers):
    max_num = 0
    second_max = 0
    
    for num in numbers:
        if num > max_num:
            second_max = max_num
            max_num = num
        elif num > second_max and num < max_num:
            second_max = num
    
    return second_max

# Example usage
numbers = [4, 12, 7, 1, 9, 3]
second_max = find_second_maximum(numbers)
print("Second maximum number is:", second_max)
```

```python
def closest_number(n):
    # Digits available on the remote
    remote_digits = {'1', '3', '5', '7', '9'}
    

    # Helper function to check if a number can be made with remote digits
    def can_make_number(num):
        return all(digit in remote_digits for digit in str(num))
    
    # Start from the number itself and increment/decrement to find the closest match
    for offset in range(len(str(n)) * 9):
        if can_make_number(n - offset):
            return n - offset
        if can_make_number(n + offset):
            return n + offset

# Test examples
print(closest_number(21))   # Output should be 19
print(closest_number(389))  # Output should be 371
```
