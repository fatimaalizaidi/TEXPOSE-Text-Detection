```python
import os
import pandas as pd
import time
from groq import Groq
```

```python
def file_to_df(file_name):
    if os.path.exists(file_name):
        df_write = pd.read_csv(file_name)
    else:
        columns = ['Author_name', 'topic', 'title', 'Abstract','year', 'Source', 'label']
        df_write = pd.DataFrame(columns=columns)
    return df_write

def data_range(df, idx):
    topics = df["topic"].values[idx]
    title = df["title"].values[idx]

    return {
             "topic" : topics,
              "title" : title
                }
```

```python
def generate_content(title):
  response = client.chat.completions.create(
      messages = [
          {
              "role": "system",
              "content": f'''I'll provide a title for research paper.Generate an abstract, Length of the abstracts should be 200-250 words. Do not add any extra text.
              Don't include the title. Just provide me an abstract
              just abstract, no heading 
              only provide just one abstract
              abstract should be in one paragraph
              this is the topic 
              
              {title}
              '''
              },
          ],
      model = "llama-3.1-8b-instant",
      temperature=1,
      max_tokens=400,
      top_p=1

  )
  abstract = response.choices[0].message.content
  return abstract
```

```python
def add_rows(df_write, data_range, generated_abstract_array):
    topic = data_range["topic"]
    abstract = generated_abstract_array
    title = data_range["title"]

    new_row = pd.DataFrame({
        'Author_name': "Llama3",
        "topic": topic,
        'title': title,
        'Abstract': abstract,
        'year': "2024",
        'Source': 'Llama3',
        'label': 'AI'
    }, index=[0])
    df_write = pd.concat([df_write, new_row], ignore_index=True)

    return df_write  # updated DataFrame
```

# **llama dataset**

```python
# every batch will save
def main(df, idx, batch):

    file_name = '../../dataset/llama_individual_data/AI_GENERATED_ABSTRACT_USING_LLAMA_TOPICWISE_35.csv'
    df_write = file_to_df(file_name)

    data_provided = data_range(df = df ,idx = idx ) # s= 0 end = 0+5
    data = generate_content(data_provided["title"])
    
    df_write = add_rows(df_write, data_provided, data)  # Update df_write with each batch
    # update to CSV after each batch
    df_write.to_csv(file_name, index=False)
```
